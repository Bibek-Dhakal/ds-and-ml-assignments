# Vector Class with Operator Overloading
This notebook implements a `Vector` class with custom attributes and operator overloading for vector addition using Python's magic dunder method `__add__`.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Operator overloading for '+'
    def __add__(self, other):
        if isinstance(other, Vector):
            return Vector(self.x + other.x, self.y + other.y)
        return NotImplemented

    # For clear string representation when printing
    def __str__(self):
        return f"Vector({self.x}, {self.y})"
```

## Demonstration and Testing
Let's create two vector instances and add them together using the standard `+` operator.

```python
# Instantiate vectors
v1 = Vector(3, 4)
v2 = Vector(1, 2)

# Perform overloaded addition
v3 = v1 + v2

# Print results
print(f"v1: {v1}")
print(f"v2: {v2}")
print(f"v1 + v2 = {v3}")
```
