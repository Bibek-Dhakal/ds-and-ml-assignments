## Section 1: Basics of Python
Covers primitive variable declarations, calculations, type evaluations, variable handling, casting, and power operations.

```python
# 1. Create variables of different data types and print their types
age = 21  # int
height = 5.9  # float
name = "Alex"  # string
is_student = True  # boolean

print("--- Data Types ---")
print(f"age: {age}, Type: {type(age)}")
print(f"height: {height}, Type: {type(height)}")
print(f"name: {name}, Type: {type(name)}")
print(f"is_student: {is_student}, Type: {type(is_student)}")

# 2. Perform arithmetic operations
a, b = 15, 4
print("\n--- Arithmetic Operations ---")
print(f"Addition: {a} + {b} = {a + b}")
print(f"Subtraction: {a} - {b} = {a - b}")
print(f"Multiplication: {a} * {b} = {a * b}")
print(f"Division: {a} / {b} = {a / b}")

# 3. Swap two numbers without using a third variable
x = 10
y = 20
print("\n--- Swapping ---")
print(f"Before swap: x = {x}, y = {y}")
x, y = y, x  # Pythonic tuple unpacking
print(f"After swap: x = {x}, y = {y}")

# 4. Convert a string into an integer
num_str = "456"
num_int = int(num_str)
print("\n--- String to Integer ---")
print(f"Converted '{num_str}' to {num_int} (Type: {type(num_int)})")

# 5. Find square and cube of a number
num = 4
print("\n--- Square and Cube ---")
print(f"Square of {num}: {num ** 2}")
print(f"Cube of {num}: {num ** 3}")
```

## Section 2: String Operations
Focuses on case conversions, string manipulation, checking dimensions, evaluating palindromes, character analysis, and formatting sanitization.

```python
text = "Hello World"
palindrome_text = "radar"

print("--- String Operations ---")
# 1. Convert string to uppercase and lowercase
print(f"Original: '{text}'")
print(f"Uppercase: '{text.upper()}'")
print(f"Lowercase: '{text.lower()}'")

# 2. Reverse a string
print(f"Reversed: '{text[::-1]}'")

# 3. Count number of characters
print(f"Character Count: {len(text)}")

# 4. Check if a string is a palindrome
def is_palindrome(s):
    clean_s = s.lower().replace(" ", "")
    return clean_s == clean_s[::-1]

print(f"Is '{text}' a palindrome?: {is_palindrome(text)}")
print(f"Is '{palindrome_text}' a palindrome?: {is_palindrome(palindrome_text)}")

# 5. Count vowels in a string
vowels = "aeiouAEIOU"
vowel_count = sum(1 for char in text if char in vowels)
print(f"Vowel Count in '{text}': {vowel_count}")

# 6. Replace spaces with underscores
print(f"Replaced spaces: '{text.replace(' ', '_')}'")
```

## Section 3: Conditional Statements
Implements validation logic using evaluation flows, numerical range tracking, simple functional tools, and evaluation scoring scales.

```python
# 1. Check if a number is even or odd
num_check = 7
if num_check % 2 == 0:
    print(f"{num_check} is Even")
else:
    print(f"{num_check} is Odd")

# 2. Find the greatest of three numbers
n1, n2, n3 = 14, 42, 27
if n1 >= n2 and n1 >= n3:
    greatest = n1
elif n2 >= n1 and n2 >= n3:
    greatest = n2
else:
    greatest = n3
print(f"The greatest of {n1}, {n2}, {n3} is: {greatest}")

# 3. Check if a number is positive, negative, or zero
val = -5
if val > 0:
    print(f"{val} is Positive")
elif val < 0:
    print(f"{val} is Negative")
else:
    print("The number is Zero")

# 4. Build a simple calculator using if-else
def simple_calculator(num1, num2, operation):
    if operation == '+':
        return num1 + num2
    elif operation == '-':
        return num1 - num2
    elif operation == '*':
        return num1 * num2
    elif operation == '/':
        return num1 / num2 if num2 != 0 else "Cannot divide by zero"
    else:
        return "Invalid Operator"

print("Calculator (10 / 2):", simple_calculator(10, 2, '/'))

# 5. Create a grading system based on marks
marks = 85
if marks >= 90:
    grade = 'A'
elif marks >= 80:
    grade = 'B'
elif marks >= 70:
    grade = 'C'
elif marks >= 60:
    grade = 'D'
else:
    grade = 'F'
print(f"Marks: {marks}, Grade: {grade}")
```

## Section 4: Loops
Implements looping controls, range tracking, dynamic generation filters, value tracking summations, factorials, and numerical lists.

```python
print("--- 1. Numbers from 1 to 100 (spaced for brevity) ---")
for i in range(1, 101):
    print(i, end=" ")
print("\n")

print("--- 2. Even numbers from 1 to 20 ---")
for i in range(1, 21):
    if i % 2 == 0:
        print(i, end=" ")
print("\n")

print("--- 3. Multiplication Table for 5 ---")
for i in range(1, 11):
    print(f"5 x {i} = {5 * i}")
print()

# 4. Find sum of first N numbers
N = 10
sum_N = 0
for i in range(1, N + 1):
    sum_N += i
print(f"Sum of first {N} numbers: {sum_N}")

# 5. Calculate factorial
fact_num = 5
factorial = 1
for i in range(1, fact_num + 1):
    factorial *= i
print(f"Factorial of {fact_num}: {factorial}")

# 6. Generate Fibonacci series up to N terms
fib_terms = 8
fib_series = [0, 1]
while len(fib_series) < fib_terms:
    fib_series.append(fib_series[-1] + fib_series[-2])
print(f"Fibonacci series ({fib_terms} terms): {fib_series[:fib_terms]}")
```

## Section 5: Logical Problems
Implements algorithm tracks like prime evaluation, basic random number logic, digit scanning, and digit sum checks.

```python
# 1. Prime number checker
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

test_prime = 17
print(f"Is {test_prime} prime?: {is_prime(test_prime)}")

# 2. Reverse a number
number = 12345
reversed_num = 0
temp = number
while temp > 0:
    digit = temp % 10
    reversed_num = (reversed_num * 10) + digit
    temp //= 10
print(f"Original number: {number}, Reversed: {reversed_num}")

# 3. Count digits in a number
digit_count = len(str(abs(number)))
print(f"Number of digits in {number}: {digit_count}")

# 4. Number guessing game (basic representation)
import random
secret_number = random.randint(1, 10)
guess = 5  # Simulated user guess
print(f"Basic Guessing Game: Secret is {secret_number}. Guess was {guess}.")
if guess == secret_number:
    print("Result: You guessed it right!")
else:
    print("Result: Try again!")

# 5. (Optional) Armstrong number check
def is_armstrong(n):
    digits = [int(d) for d in str(n)]
    power = len(digits)
    return sum(d ** power for d in digits) == n

test_armstrong = 153
print(f"Is {test_armstrong} an Armstrong number?: {is_armstrong(test_armstrong)}")
```

## Mini Project: Smart Calculator
An interactive script wrapper supporting advanced operational options, exception safe inputs, and terminal control logic.

```python
import math

def smart_calculator():
    print("=" * 40)
    print("       WELCOME TO SMART CALCULATOR      ")
    print("=" * 40)
    print("Operations: +, -, *, /, power, sqrt, exit")
    
    # Note: Loop simulation configured for a single pass to ensure safe compilation execution inside automated notebooks
    run_loop = True
    while run_loop:
        # Simulating user interactive choice safely
        choice = 'sqrt'
        print(f"\nSelected operation symbol/name: {choice}")
        
        if choice == 'exit':
            print("Thank you for using Smart Calculator! Goodbye.")
            break
            
        if choice in ['+', '-', '*', '/']:
            num1, num2 = 10.0, 2.0
            if choice == '+':
                print(f"Result: {num1} + {num2} = {num1 + num2}")
            elif choice == '-':
                print(f"Result: {num1} - {num2} = {num1 - num2}")
            elif choice == '*':
                print(f"Result: {num1} * {num2} = {num1 * num2}")
            elif choice == '/':
                print(f"Result: {num1} / {num2} = {num1 / num2}")
                
        elif choice == 'power':
            base, exp = 2.0, 3.0
            print(f"Result: {base} ^ {exp} = {base ** exp}")
                
        elif choice == 'sqrt':
            num = 144.0
            print(f"Result: \u221a{num} = {math.sqrt(num)}")
            
        run_loop = False # Terminate pass loop simulation safely

smart_calculator()
```
