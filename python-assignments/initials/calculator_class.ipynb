# Calculator Class, Inheritance, and Polymorphism
This notebook implements a basic `Calculator` base class and inherits from it to create a `ScientificCalculator` and a `ProgrammerCalculator` to demonstrate object-oriented principles.

## 1. Base Calculator Class
Contains the fundamental mathematical operations: addition, subtraction, multiplication, and division.

```python
class Calculator:
    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b

    def multiply(self, a, b):
        return a * b

    def divide(self, a, b):
        if b == 0:
            return "Error: Division by zero"
        return a / b
```

## 2. Scientific Calculator Class
Inherits from `Calculator` and adds advanced features like exponents, square roots, and scientific formatting coefficients ($10^x$).

```python
import math

class ScientificCalculator(Calculator):
    def power(self, base, exponent):
        return base ** exponent

    def square_root(self, value):
        if value < 0:
            return "Error: Cannot take square root of a negative number"
        return math.sqrt(value)

    def multiply_with_10_to_power(self, value, exponent):
        return value * (10 ** exponent)
```

## 3. Programmer Calculator Class
Inherits from `Calculator` and includes methods to toggle values between decimal and binary representations.

```python
class ProgrammerCalculator(Calculator):
    def binary_to_decimal(self, binary_str):
        try:
            return int(str(binary_str), 2)
        except ValueError:
            return "Error: Invalid binary string"

    def decimal_to_binary(self, decimal_val):
        try:
            return bin(int(decimal_val))[2:]
        except ValueError:
            return "Error: Invalid decimal integer"
```

## 4. Execution and Testing Verification
Let's test the instances to verify polymorphism and class features.

```python
print("--- Base Calculator ---")
calc = Calculator()
print("Addition (5 + 3):", calc.add(5, 3))
print("Division (10 / 2):", calc.divide(10, 2))

print("\n--- Scientific Calculator ---")
sci_calc = ScientificCalculator()
print("Inherited Add (12 + 4):", sci_calc.add(12, 4)) 
print("Power (2^5):", sci_calc.power(2, 5))
print("Square Root (81):", sci_calc.square_root(81))
print("Multiply with 10^3:", sci_calc.multiply_with_10_to_power(5.5, 3))

print("\n--- Programmer Calculator ---")
prog_calc = ProgrammerCalculator()
print("Inherited Multiply (4 * 5):", prog_calc.multiply(4, 5))
print("Binary '1010' to Decimal:", prog_calc.binary_to_decimal("1010"))
print("Decimal 15 to Binary:", prog_calc.decimal_to_binary(15))
```
