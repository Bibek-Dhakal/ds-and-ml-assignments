6. **Data Visualization:**
   - **Histogram:**
     - Create a histogram to visualize the distribution of `sepal_length` values.
   - **Scatter Plot:**
     - Create a scatter plot with `sepal_length` on the x-axis and `petal_length` on the y-axis. Use different colors to represent different species.
   - **Box Plot:**
     - Create a box plot for the `sepal_width` values, grouped by `species`.

```python
import matplotlib.pyplot as plt

# histogram for sepal length
plt.figure(figsize=(8, 6))
plt.hist(iris_df['SepalLengthCm'], bins=20, color='blue', edgecolor='black')
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# scatter plot for sepal length vs petal length
plt.figure(figsize=(8, 6))
species = iris_df['Species'].unique()
colors = ['red', 'green', 'blue']
for i, specie in enumerate(species):
    subset = iris_df[iris_df['Species'] == specie]
    plt.scatter(subset['SepalLengthCm'], subset['PetalLengthCm'], color=colors[i], label=specie)
plt.title('Sepal Length vs Petal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.legend()
plt.grid()
plt.show()
```

```python
# box plot for sepal width grouped by species
plt.figure(figsize=(8, 6))
iris_df.boxplot(column='SepalWidthCm', by='Species', grid=False)
plt.title('Box Plot of Sepal Width by Species')
plt.suptitle('')
plt.xlabel('Species')
plt.ylabel('Sepal Width (cm)')
plt.show()
```


7. **Correlation Matrix**

### **Problem 3: Handling Missing Data and Imputation**  
1. Identify columns with **missing values**.
2. Impute the missing values:
   - For **numerical columns**, replace missing values with the **median**.
   - For **categorical columns**, replace missing values with the **most frequent** value.
3. Justify your choice of imputation methods and check if any missing data remains.

```python
missing_values = house_prices_df.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Numerical columns -> fill with median
num_cols = house_prices_df.select_dtypes(include=['number']).columns
house_prices_df[num_cols] = house_prices_df[num_cols].fillna(
    house_prices_df[num_cols].median()
)

# Categorical columns -> fill with most frequent value (mode)
cat_cols = house_prices_df.select_dtypes(include=['str']).columns
house_prices_df[cat_cols] = house_prices_df[cat_cols].fillna(
    house_prices_df[cat_cols].mode().iloc[0]
)
```

```python
remaining_missing = house_prices_df.isnull().sum()
remaining_missing = remaining_missing[remaining_missing > 0]

print("\nRemaining missing values after imputation:")
print(remaining_missing)
```

### **Problem 4: Numpy Array Operations**  
1. Convert the numerical columns (such as `LotArea`, `YearBuilt`, `GrLivArea`, `SalePrice`, etc.) into a **NumPy array**.
2. Perform the following NumPy operations on the array:
   - Compute the **mean** and **standard deviation** of each column.
   - Normalize the data using **min-max scaling** and **z-score normalization**.
   - Compute the **dot product** between two numerical columns of your choice.

```python
num_df = house_prices_df.select_dtypes(include=['number'])

num_array = num_df.to_numpy()

print("NumPy Array Shape:")
print(num_array.shape)
```

```python
import numpy as np

# Mean of each column
column_means = np.mean(num_array, axis=0)

# Standard deviation of each column
column_std = np.std(num_array, axis=0)

print("\nColumn Means:")
print(column_means)

print("\nColumn Standard Deviations:")
print(column_std)
```

```python
# Min-Max Scaling
min_max_scaled = (num_array - np.min(num_array, axis=0)) / (
    np.max(num_array, axis=0) - np.min(num_array, axis=0)
)

print("\nMin-Max Scaled Data:")
print(min_max_scaled)
```

```python
# Z-Score Normalization
z_score_normalized = (num_array - np.mean(num_array, axis=0)) / np.std(num_array, axis=0)

print("\nZ-Score Normalized Data:")
print(z_score_normalized)
```

```python
# Dot Product between two numerical columns
lotarea = house_prices_df['LotArea'].to_numpy()
saleprice = house_prices_df['SalePrice'].to_numpy()

dot_product = np.dot(lotarea, saleprice)

print("\nDot Product of LotArea and SalePrice:")
print(dot_product)
```

### **Problem 6: Aggregation and Transformation**  
1. Perform a **groupby** operation on `Neighborhood` and calculate the **mean** and **median** for `SalePrice`.
2. Use the `apply()` function to create a custom transformation for `SalePrice` (e.g., scaling or transformation).
3. Aggregate multiple statistical values (e.g., **mean**, **median**, and **count**) for `SalePrice` grouped by `OverallQual`.

```python
neighborhood_stats = house_prices_df.groupby('Neighborhood')['SalePrice'].agg(
    ['mean', 'median']
)

print("Mean and Median SalePrice by Neighborhood:")
print(neighborhood_stats)
```

```python
house_prices_df['SalePrice_Scaled'] = house_prices_df['SalePrice'].apply(
    lambda x: (
        (x - house_prices_df['SalePrice'].min()) /
        (house_prices_df['SalePrice'].max() - house_prices_df['SalePrice'].min())
    )
)

print("\nScaled SalePrice:")
print(house_prices_df[['SalePrice', 'SalePrice_Scaled']].head())
```

```python
overallqual_stats = house_prices_df.groupby('OverallQual')['SalePrice'].agg(
    ['mean', 'median', 'count']
)

print("\nSalePrice Statistics by OverallQual:")
print(overallqual_stats)
```
